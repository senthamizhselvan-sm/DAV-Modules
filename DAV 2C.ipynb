```python
import pandas as pd

text_df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

temp_df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
temp_df.to_excel("iris.xlsx", index=False)

excel_df = pd.read_excel("iris.xlsx")

web_df = pd.read_csv("https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv")

print("=" * 80)
print("TEXT DATASET (Titanic)")
print("=" * 80)
print("Shape:", text_df.shape)
print("\nColumns:")
print(text_df.columns.tolist())
print("\nFirst 5 Rows:")
print(text_df.head())
print("\nLast 5 Rows:")
print(text_df.tail())
print("\nData Types:")
print(text_df.dtypes)
print("\nMissing Values:")
print(text_df.isnull().sum())
print("\nSummary Statistics:")
print(text_df.describe(include="all"))

print("\n\n" + "=" * 80)
print("EXCEL DATASET (Iris)")
print("=" * 80)
print("Shape:", excel_df.shape)
print("\nColumns:")
print(excel_df.columns.tolist())
print("\nFirst 5 Rows:")
print(excel_df.head())
print("\nLast 5 Rows:")
print(excel_df.tail())
print("\nData Types:")
print(excel_df.dtypes)
print("\nMissing Values:")
print(excel_df.isnull().sum())
print("\nSummary Statistics:")
print(excel_df.describe(include="all"))

print("\n\n" + "=" * 80)
print("WEB DATASET (Countries)")
print("=" * 80)
print("Shape:", web_df.shape)
print("\nColumns:")
print(web_df.columns.tolist())
print("\nFirst 5 Rows:")
print(web_df.head())
print("\nLast 5 Rows:")
print(web_df.tail())
print("\nData Types:")
print(web_df.dtypes)
print("\nMissing Values:")
print(web_df.isnull().sum())
print("\nSummary Statistics:")
print(web_df.describe(include="all"))

text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\n" + "=" * 80)
print("ADDITIONAL ANALYSIS")
print("=" * 80)

print("\nTitanic Passenger Class Count:")
print(text_df["Pclass"].value_counts())

print("\nTitanic Survival Count:")
print(text_df["Survived"].value_counts())

print("\nAverage Age:", text_df["Age"].mean())
print("Maximum Fare:", text_df["Fare"].max())
print("Minimum Fare:", text_df["Fare"].min())

print("\nIris Species Count:")
print(excel_df["species"].value_counts())

print("\nAverage Sepal Length:")
print(excel_df.groupby("species")["sepal_length"].mean())

print("\nCountries Dataset:")
print("Total Countries:", len(web_df))
print("Unique Regions (if available):", web_df.nunique())

print("\nFiles Saved Successfully")
print("1. processed_text.csv")
print("2. processed_excel.xlsx")
```
