```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

print("="*100)
print("TITANIC DATASET ANALYSIS")
print("="*100)

print("\nFirst 10 Rows")
print(df.head(10))

print("\nLast 10 Rows")
print(df.tail(10))

print("\nShape")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nDuplicate Rows")
print(df.duplicated().sum())

print("\nSummary Statistics")
print(df.describe(include='all'))

print("\nSurvival Count")
print(df["Survived"].value_counts())

print("\nPassenger Class Count")
print(df["Pclass"].value_counts())

print("\nGender Count")
print(df["Sex"].value_counts())

print("\nEmbarked Port Count")
print(df["Embarked"].value_counts())

print("\nAverage Fare by Passenger Class")
print(df.groupby("Pclass")["Fare"].mean())

print("\nAverage Age by Gender")
print(df.groupby("Sex")["Age"].mean())

print("\nAverage Fare by Survival")
print(df.groupby("Survived")["Fare"].mean())

print("\nCorrelation Matrix")
print(df[["Age","Fare","SibSp","Parch","Pclass","Survived"]].corr())

print("\nTop 10 Highest Fare Passengers")
print(df.nlargest(10,"Fare")[["Name","Fare","Pclass"]])

print("\nTop 10 Oldest Passengers")
print(df.nlargest(10,"Age")[["Name","Age","Sex"]])

fig, axes = plt.subplots(3,3,figsize=(20,16))

sns.countplot(data=df,x="Survived",ax=axes[0,0])
axes[0,0].set_title("Survival Count")

sns.countplot(data=df,x="Sex",ax=axes[0,1])
axes[0,1].set_title("Gender Distribution")

sns.countplot(data=df,x="Pclass",ax=axes[0,2])
axes[0,2].set_title("Passenger Class")

sns.histplot(data=df,x="Age",bins=30,kde=True,ax=axes[1,0])
axes[1,0].set_title("Age Distribution")

sns.histplot(data=df,x="Fare",bins=30,kde=True,ax=axes[1,1])
axes[1,1].set_title("Fare Distribution")

sns.boxplot(data=df,x="Pclass",y="Fare",ax=axes[1,2])
axes[1,2].set_title("Fare by Passenger Class")

sns.scatterplot(data=df,x="Age",y="Fare",hue="Survived",ax=axes[2,0])
axes[2,0].set_title("Age vs Fare")

sns.violinplot(data=df,x="Survived",y="Age",ax=axes[2,1])
axes[2,1].set_title("Age vs Survival")

sns.barplot(data=df,x="Pclass",y="Fare",estimator="mean",errorbar=None,ax=axes[2,2])
axes[2,2].set_title("Average Fare by Class")

plt.tight_layout()
plt.show()

plt.figure(figsize=(8,6))
sns.heatmap(df[["Age","Fare","SibSp","Parch","Pclass","Survived"]].corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

pair_df = df[["Age","Fare","Pclass","Survived"]].dropna()

sns.pairplot(pair_df,hue="Survived")
plt.show()

print("\nMaximum Values")
print(df[["Age","Fare","SibSp","Parch"]].max())

print("\nMinimum Values")
print(df[["Age","Fare","SibSp","Parch"]].min())

print("\nMean Values")
print(df[["Age","Fare","SibSp","Parch"]].mean())

print("\nMedian Values")
print(df[["Age","Fare","SibSp","Parch"]].median())

print("\nStandard Deviation")
print(df[["Age","Fare","SibSp","Parch"]].std())

print("\nVariance")
print(df[["Age","Fare","SibSp","Parch"]].var())

print("\nAnalysis Completed Successfully")
```
