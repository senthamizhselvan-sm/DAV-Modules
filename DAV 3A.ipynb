```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

df = pd.read_csv("https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv")

df.columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

print("=" * 100)
print("PIMA INDIANS DIABETES DATASET")
print("=" * 100)

print("\nFirst 10 Rows")
print(df.head(10))

print("\nLast 10 Rows")
print(df.tail(10))

print("\nShape")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nSummary Statistics")
print(df.describe())

numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

def univariate_analysis(dataframe, columns):
    stats = {}
    for col in columns:
        stats[col] = {
            "Mean": dataframe[col].mean(),
            "Median": dataframe[col].median(),
            "Mode": dataframe[col].mode()[0],
            "Minimum": dataframe[col].min(),
            "Maximum": dataframe[col].max(),
            "Range": dataframe[col].max() - dataframe[col].min(),
            "Variance": dataframe[col].var(),
            "Standard Deviation": dataframe[col].std(),
            "Skewness": skew(dataframe[col]),
            "Kurtosis": kurtosis(dataframe[col])
        }
    return pd.DataFrame(stats).T

statistics = univariate_analysis(df, numerical_columns)

print("\nUnivariate Analysis")
print(statistics)

print("\nOutcome Count")
print(df["Outcome"].value_counts())

print("\nCorrelation Matrix")
print(df[numerical_columns + ["Outcome"]].corr())

print("\nAnalysis Completed Successfully")
```
