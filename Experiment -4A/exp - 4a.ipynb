```python
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis

# --------------------------------------------------
# 1. Load Pima Indians Diabetes Dataset
# --------------------------------------------------

pima_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome"
]

pima_diabetes = pd.read_csv(
    "pima-indians-diabetes.csv",
    header=None,
    names=pima_columns
)


# --------------------------------------------------
# 2. Load UCI Diabetes Dataset
# --------------------------------------------------

uci_diabetes = pd.read_csv(
    "diabetes_merged_date-time-sorted-includes-patient-id.csv"
)


# --------------------------------------------------
# 3. Display Dataset Samples
# --------------------------------------------------

print("Pima Indians Diabetes Dataset Sample:")
print(pima_diabetes.head())

print("\nUCI Diabetes Dataset Sample:")
print(uci_diabetes.head())


# --------------------------------------------------
# 4. Display Column Names
# --------------------------------------------------

print("\nPima Dataset Columns:")
print(pima_diabetes.columns.tolist())

print("\nUCI Dataset Columns:")
print(uci_diabetes.columns.tolist())


# --------------------------------------------------
# 5. Define Numerical Columns
# --------------------------------------------------

pima_numerical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

uci_numerical_columns = [
    "patient_id",
    "code",
    "value"
]


# --------------------------------------------------
# 6. Univariate Analysis Function
# --------------------------------------------------

def univariate_analysis(df, columns):

    stats = {}

    for col in columns:

        stats[col] = {
            "Mean": np.mean(df[col]),
            "Median": np.median(df[col]),
            "Mode": df[col].mode()[0],
            "Variance": np.var(df[col], ddof=1),
            "Standard Deviation": np.std(df[col], ddof=1),
            "Skewness": skew(df[col]),
            "Kurtosis": kurtosis(df[col])
        }

    return pd.DataFrame(stats).T


# --------------------------------------------------
# 7. Perform Univariate Analysis
# --------------------------------------------------

pima_stats = univariate_analysis(
    pima_diabetes,
    pima_numerical_columns
)

uci_stats = univariate_analysis(
    uci_diabetes,
    uci_numerical_columns
)


# --------------------------------------------------
# 8. Display Results
# --------------------------------------------------

print("\nPima Indians Diabetes Dataset Statistics:")
print(pima_stats)

print("\nUCI Diabetes Dataset Statistics:")
print(uci_stats)
```
